```sql
%python
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()

CATALOG = "credit_analysis_catalog"


# GOLD - DIM APPLICANT

def test_dim_applicant_not_empty():

    df = spark.table(
        f"{CATALOG}.gold.dim_applicant"
    )

    assert df.count() > 0, \
        "dim_applicant table is empty"


def test_dim_applicant_key_not_null():

    df = spark.table(
        f"{CATALOG}.gold.dim_applicant"
    )

    null_count = df.filter(
        "applicant_key IS NULL"
    ).count()

    assert null_count == 0, \
        f"Found {null_count} null applicant_key values"


def test_dim_applicant_key_unique():

    df = spark.table(
        f"{CATALOG}.gold.dim_applicant"
    )

    duplicate_count = (
        df.groupBy("applicant_key")
        .count()
        .filter("count > 1")
        .count()
    )

    assert duplicate_count == 0, \
        f"Found {duplicate_count} duplicate applicant_key groups"


def test_dim_applicant_id_unique():

    df = spark.table(
        f"{CATALOG}.gold.dim_applicant"
    )

    duplicate_count = (
        df.groupBy("applicant_id")
        .count()
        .filter("count > 1")
        .count()
    )

    assert duplicate_count == 0, \
        f"Found {duplicate_count} duplicate applicant_id groups"


# GOLD - DIM CREDIT PROFILE

def test_dim_credit_profile_not_empty():

    df = spark.table(
        f"{CATALOG}.gold.dim_credit_profile"
    )

    assert df.count() > 0, \
        "dim_credit_profile table is empty"


def test_dim_credit_profile_key_not_null():

    df = spark.table(
        f"{CATALOG}.gold.dim_credit_profile"
    )

    null_count = df.filter(
        "credit_profile_key IS NULL"
    ).count()

    assert null_count == 0, \
        f"Found {null_count} null credit_profile_key values"


def test_dim_credit_profile_key_unique():

    df = spark.table(
        f"{CATALOG}.gold.dim_credit_profile"
    )

    duplicate_count = (
        df.groupBy("credit_profile_key")
        .count()
        .filter("count > 1")
        .count()
    )

    assert duplicate_count == 0, \
        f"Found {duplicate_count} duplicate credit_profile_key groups"


def test_dim_credit_profile_applicant_id_unique():

    df = spark.table(
        f"{CATALOG}.gold.dim_credit_profile"
    )

    duplicate_count = (
        df.groupBy("applicant_id")
        .count()
        .filter("count > 1")
        .count()
    )

    assert duplicate_count == 0, \
        f"Found {duplicate_count} duplicate applicant_id groups"


# GOLD - DIM ECONOMIC

def test_dim_economic_not_empty():

    df = spark.table(
        f"{CATALOG}.gold.dim_economic"
    )

    assert df.count() > 0, \
        "dim_economic table is empty"


def test_dim_economic_key_not_null():

    df = spark.table(
        f"{CATALOG}.gold.dim_economic"
    )

    null_count = df.filter(
        "economic_key IS NULL"
    ).count()

    assert null_count == 0, \
        f"Found {null_count} null economic_key values"


def test_dim_economic_key_unique():

    df = spark.table(
        f"{CATALOG}.gold.dim_economic"
    )

    duplicate_count = (
        df.groupBy("economic_key")
        .count()
        .filter("count > 1")
        .count()
    )

    assert duplicate_count == 0, \
        f"Found {duplicate_count} duplicate economic_key groups"


def test_dim_economic_year_region_unique():

    df = spark.table(
        f"{CATALOG}.gold.dim_economic"
    )

    duplicate_count = (
        df.groupBy("year", "region")
        .count()
        .filter("count > 1")
        .count()
    )

    assert duplicate_count == 0, \
        f"Found {duplicate_count} duplicate (year, region) groups"


# GOLD - DIM LOAN

def test_dim_loan_not_empty():

    df = spark.table(
        f"{CATALOG}.gold.dim_loan"
    )

    assert df.count() > 0, \
        "dim_loan table is empty"


def test_dim_loan_key_not_null():

    df = spark.table(
        f"{CATALOG}.gold.dim_loan"
    )

    null_count = df.filter(
        "loan_key IS NULL"
    ).count()

    assert null_count == 0, \
        f"Found {null_count} null loan_key values"


def test_dim_loan_key_unique():

    df = spark.table(
        f"{CATALOG}.gold.dim_loan"
    )

    duplicate_count = (
        df.groupBy("loan_key")
        .count()
        .filter("count > 1")
        .count()
    )

    assert duplicate_count == 0, \
        f"Found {duplicate_count} duplicate loan_key groups"


def test_dim_loan_id_unique():

    df = spark.table(
        f"{CATALOG}.gold.dim_loan"
    )

    duplicate_count = (
        df.groupBy("loan_id")
        .count()
        .filter("count > 1")
        .count()
    )

    assert duplicate_count == 0, \
        f"Found {duplicate_count} duplicate loan_id groups"


# GOLD - FACT CREDIT APPLICATION

def test_fact_credit_application_not_empty():

    df = spark.table(
        f"{CATALOG}.gold.fact_credit_application"
    )

    assert df.count() > 0, \
        "fact_credit_application table is empty"


def test_fact_credit_application_key_not_null():

    df = spark.table(
        f"{CATALOG}.gold.fact_credit_application"
    )

    null_count = df.filter(
        "application_key IS NULL"
    ).count()

    assert null_count == 0, \
        f"Found {null_count} null application_key values"


def test_fact_credit_application_key_unique():

    df = spark.table(
        f"{CATALOG}.gold.fact_credit_application"
    )

    duplicate_count = (
        df.groupBy("application_key")
        .count()
        .filter("count > 1")
        .count()
    )

    assert duplicate_count == 0, \
        f"Found {duplicate_count} duplicate application_key groups"


def test_fact_credit_application_applicant_id_not_null():

    df = spark.table(
        f"{CATALOG}.gold.fact_credit_application"
    )

    null_count = df.filter(
        "applicant_id IS NULL"
    ).count()

    assert null_count == 0, \
        f"Found {null_count} null applicant_id values"


def test_fact_credit_application_applicant_key_not_null():

    df = spark.table(
        f"{CATALOG}.gold.fact_credit_application"
    )

    null_count = df.filter(
        "applicant_key IS NULL"
    ).count()

    assert null_count == 0, \
        f"Found {null_count} null applicant_key values"


def test_fact_credit_application_no_duplicate_applicants():

    df = spark.table(
        f"{CATALOG}.gold.fact_credit_application"
    )

    duplicate_count = (
        df.groupBy("applicant_id")
        .count()
        .filter("count > 1")
        .count()
    )

    assert duplicate_count == 0, \
        f"Found {duplicate_count} duplicate applicant_id groups"


def test_fact_credit_application_loan_amount_non_negative():

    df = spark.table(
        f"{CATALOG}.gold.fact_credit_application"
    )

    invalid_count = df.filter(
        "loan_amount < 0"
    ).count()

    assert invalid_count == 0, \
        f"Found {invalid_count} negative loan_amount values"


# RUN ALL GOLD TESTS

test_functions = [

    # dim_applicant
    test_dim_applicant_not_empty,
    test_dim_applicant_key_not_null,
    test_dim_applicant_key_unique,
    test_dim_applicant_id_unique,

    # dim_credit_profile
    test_dim_credit_profile_not_empty,
    test_dim_credit_profile_key_not_null,
    test_dim_credit_profile_key_unique,
    test_dim_credit_profile_applicant_id_unique,

    # dim_economic
    test_dim_economic_not_empty,
    test_dim_economic_key_not_null,
    test_dim_economic_key_unique,
    test_dim_economic_year_region_unique,

    # dim_loan
    test_dim_loan_not_empty,
    test_dim_loan_key_not_null,
    test_dim_loan_key_unique,
    test_dim_loan_id_unique,

    # fact_credit_application
    test_fact_credit_application_not_empty,
    test_fact_credit_application_key_not_null,
    test_fact_credit_application_key_unique,
    test_fact_credit_application_applicant_id_not_null,
    test_fact_credit_application_applicant_key_not_null,
    test_fact_credit_application_no_duplicate_applicants,
    test_fact_credit_application_loan_amount_non_negative,
]


# TEST RUNNER

passed = 0
failed = 0

for test in test_functions:

    try:

        test()

        print(f"PASSED: {test.__name__}")

        passed += 1

    except AssertionError as e:

        print(f"FAILED: {test.__name__}")
        print(f"       {e}")

        failed += 1

    except Exception as e:

        print(f"ERROR: {test.__name__}")
        print(f"       {e}")

        failed += 1


print("\n" + "=" * 60)
print("GOLD TEST SUMMARY")
print("=" * 60)

print(f"Total Tests : {len(test_functions)}")
print(f"Passed      : {passed}")
print(f"Failed      : {failed}")

success_rate = (
    passed / len(test_functions) * 100
    if len(test_functions) > 0
    else 0
)

print(f"Success %   : {success_rate:.2f}%")
```
