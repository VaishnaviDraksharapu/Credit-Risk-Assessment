```sql
%python
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()

CATALOG = "credit_analysis_catalog"


# BRONZE - APPLICANT PROFILES

def test_bronze_applicant_profiles_not_empty():

    df = spark.table(
        f"{CATALOG}.bronze.applicant_profiles"
    )

    assert df.count() > 0, \
        "Bronze applicant_profiles table is empty"


def test_bronze_applicant_profiles_applicant_id_not_null():

    df = spark.table(
        f"{CATALOG}.bronze.applicant_profiles"
    )

    null_count = df.filter(
        "applicant_id IS NULL"
    ).count()

    assert null_count == 0, \
        f"Found {null_count} null applicant_id values"


# BRONZE - CREDIT APPLICATION

def test_bronze_credit_application_not_empty():

    df = spark.table(
        f"{CATALOG}.bronze.credit_application"
    )

    assert df.count() > 0, \
        "Bronze credit_application table is empty"


def test_bronze_credit_application_applicant_id_not_null():

    df = spark.table(
        f"{CATALOG}.bronze.credit_application"
    )

    null_count = df.filter(
        "applicant_id IS NULL"
    ).count()

    assert null_count == 0, \
        f"Found {null_count} null applicant_id values"


# BRONZE - CREDIT HISTORY

def test_bronze_credit_history_not_empty():

    df = spark.table(
        f"{CATALOG}.bronze.credit_history"
    )

    assert df.count() > 0, \
        "Bronze credit_history table is empty"


def test_bronze_credit_history_applicant_id_not_null():

    df = spark.table(
        f"{CATALOG}.bronze.credit_history"
    )

    null_count = df.filter(
        "applicant_id IS NULL"
    ).count()

    assert null_count == 0, \
        f"Found {null_count} null applicant_id values"


# BRONZE - LOAN DETAILS

def test_bronze_loan_details_not_empty():

    df = spark.table(
        f"{CATALOG}.bronze.loan_details"
    )

    assert df.count() > 0, \
        "Bronze loan_details table is empty"


def test_bronze_loan_details_loan_id_not_null():

    df = spark.table(
        f"{CATALOG}.bronze.loan_details"
    )

    null_count = df.filter(
        "loan_id IS NULL"
    ).count()

    assert null_count == 0, \
        f"Found {null_count} null loan_id values"


# BRONZE - ECONOMIC INDICATORS

def test_bronze_economic_indicators_not_empty():

    df = spark.table(
        f"{CATALOG}.bronze.economic_indicators"
    )

    assert df.count() > 0, \
        "Bronze economic_indicators table is empty"


def test_bronze_economic_indicators_year_not_null():

    df = spark.table(
        f"{CATALOG}.bronze.economic_indicators"
    )

    null_count = df.filter(
        "year IS NULL"
    ).count()

    assert null_count == 0, \
        f"Found {null_count} null year values"


def test_bronze_economic_indicators_region_not_null():

    df = spark.table(
        f"{CATALOG}.bronze.economic_indicators"
    )

    null_count = df.filter(
        "region IS NULL"
    ).count()

    assert null_count == 0, \
        f"Found {null_count} null region values"


# RUN ALL BRONZE TESTS

test_functions = [
    test_bronze_applicant_profiles_not_empty,
    test_bronze_applicant_profiles_applicant_id_not_null,

    test_bronze_credit_application_not_empty,
    test_bronze_credit_application_applicant_id_not_null,

    test_bronze_credit_history_not_empty,
    test_bronze_credit_history_applicant_id_not_null,

    test_bronze_loan_details_not_empty,
    test_bronze_loan_details_loan_id_not_null,

    test_bronze_economic_indicators_not_empty,
    test_bronze_economic_indicators_year_not_null,
    test_bronze_economic_indicators_region_not_null,
]


passed = 0
failed = 0

for test in test_functions:

    try:

        test()

        print(f"PASSED: {test.__name__}")

        passed += 1

    except AssertionError as e:

        print(f"FAILED: {test.__name__}")

        print(f"       {e}")

        failed += 1

    except Exception as e:

        print(f"ERROR: {test.__name__}")

        print(f"       {e}")

        failed += 1


print("\n" + "=" * 50)
print("BRONZE TEST SUMMARY")
print("=" * 50)

print(f"Total Tests : {len(test_functions)}")
print(f"Passed      : {passed}")
print(f"Failed      : {failed}")
print(f"Success %   : {(passed / len(test_functions)) * 100:.2f}%")
```
